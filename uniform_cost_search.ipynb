```python
def  uniform_cost_search(start, goal, graph, cost):
    answer = {node: {'cost': float('inf'), 'path': []} for node in goal}
#     **************************************
#     ******** Write your code here ********
#     **************************************
    
    # A queue of that will hold 3 things per entry:
    # [(cost to next node, next node, path to get to next node)]
    priority_queue = [(0, start, [start])]

    # A set to see if our search has been to any node before
    visited = set()

    # While the priority_queue is not empty
    while priority_queue:

        # Sort the priority_queue by cost of next node
        priority_queue.sort()

        # print (priority_queue)

        # Get data from the top (aka lowest cost) node
        current_cost = priority_queue[0][0]
        current_node = priority_queue[0][1]
        path = priority_queue[0][2]
        # Remove node now that we have the data
        priority_queue.pop(0)

        # Mark the new node as visited so we don't come back
        visited.add(current_node)

        # print (f"path {path}")
        # print (f"{current_node}   {graph[current_node]}")

        # If we have reached our goal
        if current_node in goal:
            # Set values of answer to the current node
            answer[current_node]['cost'] = current_cost
            answer[current_node]['path'] = path

            # If the goal has multiple states
            if (len(goal) > 1):
                # Get rid of the node we've discovered
                goal.remove(current_node)
            # If there was only 1 goal state                
            else:
                return answer
        
        # For the current node, check all nodes it is adjacent to
        for neighbor in graph[current_node]:
            # Find the cost of edge between node and neighbor
            edge_cost = cost[ ( current_node, neighbor) ]
            # If that neighbor hasn't been visited yet
            if neighbor not in visited:
                # We add to the priority_queue 
                # the total cost of traversing to the neighbor, the neighbor, and the path to that neighbor 
                priority_queue.append(( current_cost + edge_cost, neighbor, path + [neighbor]))

    # If the goal was never found, then that node is not on the given graph
    if (goal != []):
        print (f"{goal} was not found!")
    
    return answer
 
# main function
if __name__ == '__main__':
     
    # create a graph with no more than 30 nodes
    graph, cost = [[] for i in range(30)], {}
 
    # add edges to the graph
    graph[0].append(4)
    graph[0].append(5)
    graph[0].append(16)
    graph[2].append(1)
    graph[3].append(1)
    graph[4].append(2)
    graph[4].append(3)
    graph[4].append(5)
    graph[5].append(8)
    graph[5].append(18)
    graph[6].append(3)
    graph[6].append(7)
    graph[8].append(16)
    graph[8].append(17)
    graph[16].append(17)
    graph[18].append(6)
    
 
    # add cost to each edge
    cost[(0, 4)] = 3
    cost[(0, 5)] = 9
    cost[(0, 16)] = 1
    cost[(2, 1)] = 2
    cost[(3, 1)] = 2
    cost[(4, 2)] = 1
    cost[(4, 3)] = 8
    cost[(4, 5)] = 2
    cost[(5, 8)] = 3
    cost[(5, 18)] = 2
    cost[(6, 3)] = 3
    cost[(6, 7)] = 2
    cost[(8, 16)] = 4
    cost[(8, 17)] = 4
    cost[(16, 17)] = 15
    cost[(18, 6)] = 1
    
    # set start state 
    start = 0
    
    # set goal state, there can be multiple goal states
    goal = [7]
    
    # call uniform_search_cost function to get the minimum cost to reach the goal and the path with minumum cost
    # ****** You have to implement this function *****
    min_cost_info = uniform_cost_search(start, goal, graph, cost)

    for node, info in min_cost_info.items():
        print(f'Minimum cost from {start} to {node} is {info["cost"]}')
        print(f'Path: {info["path"]}')
```
